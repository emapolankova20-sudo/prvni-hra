```python

def volba_hrace(pocet):
    while True:
        try:
            volba = int(input("Vyber číslo: "))

            if volba >= 1 and volba <= pocet:
                return volba
            else:
                print("Vyber správné číslo!")

        except ValueError:
            print("Musíš zadat číslo!")


def ukaz_inventar(inventar):
    print("---------- INVENTÁŘ ----------")

    if len(inventar) == 0:
        print("Tvůj inventář je prázdný.")
    else:
        for predmet in inventar:
            print("-", predmet)

    print("------------------------------")

def hra():
    inventar = []

    print("****************************************")
    print("           POSLEDNÍ POLIBEK")
    print("****************************************")

    jmeno = input("Jak se jmenuje hlavní hrdinka? ")

    while jmeno == "":
        print("Musíš zadat jméno.")
        jmeno = input("Jak se jmenuje hlavní hrdinka? ")

    videla_zenu, zavolala_policii = uvod()

    alexovi_veri = setkani_s_alexem()

    varovala_kamaradka = pozvani_do_domu()

    alex_ma_nuz, nasla_varovani = dum(
        inventar, videla_zenu, zavolala_policii, varovala_kamaradka)

    alex_ma_nuz = sklep(inventar, alex_ma_nuz, zavolala_policii)

    konec_hry(inventar, alex_ma_nuz, zavolala_policii)

    print("Děkuji za hraní " + jmeno + "!")


def uvod():
    print("Je pátek večer a vracíš se sama domů.")
    print("Ulice jsou prázdné a začíná pršet.")
    print("Najednou uslyšíš ženský křik z vedlejší uličky.")

    print("1 - Půjdeš se podívat")
    print("2 - Budeš pokračovat domů")
    print("3 - Zavoláš policii")

    volba = volba_hrace(3)

    videla_zenu = False
    zavolala_policii = False

    if volba == 1:
        videla_zenu = True
        print("V uličce najdeš ženu ležící na zemi.")
        print("Vedle ní je velká krvavá skvrna.")
        print("Žena už nereaguje.")
        print("Najednou za sebou uslyšíš kroky.")

    elif volba == 2:
        print("Rychle odejdeš.")
        print("Když se otočíš, na konci ulice stojí neznámý muž.")
        print("Dívá se přímo na tebe.")

    else:
        zavolala_policii = True
        print("Zavoláš policii.")
        print("Když přijedou, v uličce už nikdo není.")
        print("Policista ti řekne, že se ti to možná jen zdálo.")

    return videla_zenu, zavolala_policii


def setkani_s_alexem():
    print("Najednou k tobě přijde tajemný kluk.")
    print("ALEX: Ahoj, neměla bys tu být sama.")
    print("1 - Dáš se s ním do řeči")
    print("2 - Odejdeš")
    print("3 - Zeptáš se, jestli něco neví o té ženě")

    volba = volba_hrace(3)
    alexovi_veri = False

    if volba == 1:
        alexovi_veri = True
        print("Alex se usměje a nabídne ti doprovod domů.")
    elif volba == 2:
        print("Alex tě chvíli sleduje a potom odejde.")
    else:
        print("Alex zvážní.")
        print("ALEX: O některých věcech je lepší nevědět.")

    return alexovi_veri


def pozvani_do_domu():
    print("O několik dní později ti Alex napíše.")
    print("Pozve tě do starého domu na okraji města.")
    print("1 - Půjdeš za ním")
    print("2 - Odmítneš")
    print("3 - Zeptáš se kamarádky, jestli o něm něco neví")

    volba = volba_hrace(3)
    varovala_kamaradka = False

    if volba == 1:
        print("Dorazíš k obrovskému starému domu.")
        print("Alex tě čeká před dveřmi.")
    elif volba == 2:
        print("Alex ti napíše: Stejně vím, kde bydlíš.")
        print("O několik hodin později někdo zaklepe na okno.")
        print("Poté se probudíš před starým domem.")
    else:
        varovala_kamaradka = True
        print("Kamarádka ti řekne, že Alexe zná.")
        print("Prý před několika lety záhadně zemřel.")
        print("Rozhodneš se přesto zjistit pravdu.")

    return varovala_kamaradka


def dum(inventar, videla_zenu, zavolala_policii, varovala_kamaradka):
    print("Vstoupíš do domu.")
    print("Všude je ticho a na zemi leží prach.")
    print("Na stolku najdeš baterku, rezavý klíč a starý dopis.")

    print("1 - Vezmeš si všechny předměty")
    print("2 - Vezmeš si pouze klíč")
    print("3 - Vezmeš si pouze baterku")
    print("4 - Nevezmeš si nic")

    volba = volba_hrace(4)

    if volba == 1:
        inventar.append("Baterka")
        inventar.append("Rezavý klíč")
        inventar.append("Starý dopis")
    elif volba == 2:
        inventar.append("Rezavý klíč")
    elif volba == 3:
        inventar.append("Baterka")

    print("Získala jsi nové předměty.")
    ukaz_inventar(inventar)

    print("Na stěně visí staré fotografie.")
    print("Na jedné z nich poznáš Alexe.")
    print("Fotografie je stará 20 let.")
    print("Pod ní je napsáno: ZEMŘEL V ROCE 2006.")

    print("1 - Zeptáš se Alexe na pravdu")
    print("2 - Pokusíš se utéct")
    print("3 - Budeš prohledávat dům")

    volba = volba_hrace(3)
    pokusila_se_utect = False

    if volba == 1:
        print("Alex přizná, že je mrtvý.")
        print("Tvrdí ale, že se vrátil kvůli tobě.")
    elif volba == 2:
        pokusila_se_utect = True
        print("Dveře jsou zamčené.")
        print("Za sebou uslyšíš Alexův hlas.")
        print("ALEX: Nikdy mě neopustíš!")
    else:
        print("V jedné místnosti najdeš staré noviny.")
        print("Na titulní stránce je fotografie zavražděné ženy.")

        if videla_zenu:
            print("Je to stejná žena, kterou jsi viděla v uličce.")
        else:
            print("Poznáš ženu z příběhu, o kterém jsi slyšela.")

        if zavolala_policii:
            print("V novinách je také zmínka o policejním vyšetřování.")

        print("Najdeš také seznam dalších tří obětí.")

    print("Najednou uslyšíš hlasité bouchnutí.")
    print("Z vedlejší místnosti se ozve ženský hlas.")
    print("1 - Půjdeš se podívat")
    print("2 - Schováš se")
    print("3 - Zavoláš na Alexe")

    volba = volba_hrace(3)
    alex_ma_nuz = False
    nasla_varovani = False

    if volba == 1:
        print("Otevřeš dveře.")
        print("Na zemi leží muž a vedle něj je nůž.")
        print("Muž se ještě hýbe.")
        print("1 - Pokusíš se mu pomoct")
        print("2 - Utečeš z místnosti")

        volba = volba_hrace(2)

        if volba == 1:
            nasla_varovani = True
            print("Muž tě chytí za ruku.")
            print("Zašeptá: Alex je vrah...")
            print("Potom zavře oči.")
        else:
            print("Zavřeš dveře a utečeš.")
            print("Za nimi uslyšíš poslední výkřik.")

    elif volba == 2:
        print("Schováš se za skříň.")
        print("Dveře se pomalu otevřou.")
        print("Někdo vstoupí do místnosti.")
        print("Vidíš pouze jeho boty.")
        print("Po chvíli odejde.")

    else:
        alex_ma_nuz = True
        print("Alex se na tebe podívá.")
        print("ALEX: Neměla jsi sem chodit.")
        print("Rychle utečeš do vedlejší chodby.")
        print("Za chvíli najdeš dveře do starého sklepa.")

    if pokusila_se_utect:
        print("Alex si všiml, že ses pokusila utéct.")

    if varovala_kamaradka:
        print("Vzpomeneš si na varování kamarádky.")

    if nasla_varovani:
        print("Teď už víš, že Alex je opravdu nebezpečný.")

    return alex_ma_nuz, nasla_varovani


def sklep(inventar, alex_ma_nuz, zavolala_policii):
    print("Na konci chodby najdeš dveře do starého sklepa.")
    print("Na dveřích je zámek s číselným kódem.")

    if "Rezavý klíč" in inventar:
        print("Vezmeš rezavý klíč a odemkneš dveře.")
        inventar.remove("Rezavý klíč")
    else:
        print("Nemáš klíč. Zkusíš dveře otevřít silou.")
        print("Dveře se po chvíli otevřou, ale uděláš velký hluk.")
        alex_ma_nuz = True

    print("Vstoupíš do tmavého sklepa.")
    print("Vedle schodů leží starý vzkaz.")

    if "Baterka" in inventar:
        print("Zapneš baterku a posvítíš si na stěnu.")
        print("Objevíš na ní napsaný číselný kód: 2006.")
    else:
        print("Ve sklepě je skoro úplná tma.")
        print("Na stěně nic nevidíš.")

    if "Starý dopis" in inventar:
        print("Otevřeš starý dopis.")
        print("Je v něm napsáno: Alex zemřel v roce 2006.")
        print("Pod textem je připsaný kód ke schránce: 2006.")
    else:
        print("Bez dopisu nevíš, jaký rok hledat.")
        print("Budeš muset kód zkusit odhadnout.")

    print("1 - Prohlédneš si inventář")
    print("2 - Zkusíš otevřít kovovou schránku")

    volba = volba_hrace(2)

    if volba == 1:
        ukaz_inventar(inventar)
        print("Přistoupíš ke kovové schránce.")
    else:
        print("Přistoupíš přímo ke kovové schránce.")

    print("Zadej čtyřmístný kód:")

    while True:
        try:
            kod = int(input("Kód: "))

            if kod == 2006:
                print("Správný kód!")
                print("Otevře se stará kovová schránka.")
                break
            else:
                print("Špatný kód. Zkus to znovu.")

        except ValueError:
            print("Musíš zadat číslo!")

    print("Ve schránce najdeš fotografie všech obětí.")
    print("Úplně dole je tvoje fotografie.")
    print("Pod ní je napsáno: DALŠÍ OBĚŤ.")

    if zavolala_policii:
        print("V dálce zaslechneš policejní sirény.")

    if alex_ma_nuz:
        print("Za dveřmi sklepa uslyšíš Alexovy kroky.")

    return alex_ma_nuz


def konec_hry(inventar, alex_ma_nuz, zavolala_policii):
    print("Najednou se za tebou ozvou kroky.")
    print("Alex tě nakonec najde.")
    print("ALEX: Miluji tě. Zůstaň se mnou navždy.")
    print("Vytáhne starý prsten a položí ho před tebe.")

    if alex_ma_nuz:
        print("Všimneš si, že Alex stále drží zakrvácený nůž.")

    print("Co uděláš?")
    print("1 - Zůstaneš s Alexem")
    print("2 - Pokusíš se ho zastavit")
    print("3 - Utečeš")

    volba = volba_hrace(3)

    if volba == 1:
        print("KONEC: TEMNÁ POSEDLOST")
        print("Rozhodneš se zůstat s Alexem.")
        print("Alex tě zamkne v domě.")
        print("Postupně zjistíš, že všechny jeho bývalé lásky")
        print("skončily stejně jako ženy ze seznamu.")

    elif volba == 2:
        print("Alex se na tebe vrhne.")
        print("1 - Udeříš ho")
        print("2 - Pokusíš se s ním domluvit")
        print("3 - Vezmeš ze stolu nůž")

        volba = volba_hrace(3)

        if volba == 1:
            print("Alex spadne na zem.")
            print("Využiješ chvíle a utečeš.")
            print("KONEC: POSLEDNÍ ŠANCE")
            print("Přežila jsi a policie později najde dům.")

        elif volba == 2:
            print("Alex se na chvíli uklidní.")
            print("KONEC: ZRADA")
            print("Alex předstírá, že ti odpustil.")
            print("Potom zhasne světlo.")
            print("Uslyšíš jeho kroky za sebou.")

        else:
            print("Alex se zastaví.")
            print("Pomalu ustoupí a pustí tě ven.")
            print("KONEC: POSLEDNÍ OBĚŤ")
            print("Podaří se ti zachránit si život.")
            print("Alex už ale nikdy nebyl nalezen.")

    else:
        print("Rozběhneš se směrem k tajnému tunelu.")
        print("Za sebou slyšíš Alexovy kroky.")
        print("1 - Budeš pokračovat v běhu")
        print("2 - Schováš se v jedné z místností")

        volba = volba_hrace(2)

        if volba == 1:
            print("KONEC: ÚTĚK")
            print("Podaří se ti dostat ven.")

            if zavolala_policii:
                print("Venku už čeká policie, kterou jsi zavolala.")
            else:
                print("O několik dní později dostaneš dopis.")
                print("Je na něm napsáno: JEŠTĚ SE SETKÁME.")

        else:
            print("KONEC: SKRYTÁ MÍSTNOST")
            print("Schováš se ve staré místnosti.")
            print("Najdeš tam fotografie všech obětí.")
            print("Na poslední fotografii jsi ty.")
            print("Dveře se pomalu začnou otevírat.")

    ukaz_inventar(inventar)


hra()
```
